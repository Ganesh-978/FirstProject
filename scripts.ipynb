```python
def sumOfTheDigitsOfHarshadNumber(x):
    res=0
    while x!=0:
        res=res+x%10
        x=x//10
        print(res)
    if x%res==0:
        return res
    else:
        return -1
```

```python
def isPalindrome(self, x: int) -> bool:
        if x<0:
            return False
        res,temp=0,x
        while x:
            res =res*10+x%10
            x=x//20

        return res==temp
```

```python
#2.global variables
x=5
def change():
    global x
    x=x+10
    print(x)#15
change()
print(x)#15
```

```python
#2.global variables
x=5
def change():
    x=10
    x=x+10
    print(x)#20
change()
print(x)#5
```

```python
#3.Nonlocal
def outer():
    x='local'
    def inner():
        nonlocal x
        x='non local'
        print(x)#nonlocal
    inner()
    print(x)#nonlocal
outer()
```

```python
#slicing
#    0 1 2 3 4
arr=[1,2,3,4,5]
#   -5-4-3-2-1
#arr[start,stop,step]
#default start-0,stop-len(arr),step-1
print(arr[2:])#[3,2,4]
print(arr[:2])#[1,2]
print(arr[2:4])#[3,2]
print(arr[::2])#[1,3,4]
print(arr[-4:])#[2,3,2,4]
print(arr[:-4])#[1]
print(arr[::-1])#[4,2,3,2,1]
print(arr.index(2))#1
```
